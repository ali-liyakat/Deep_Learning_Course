```python
q1 = torch.tensor([
    [200, 220, 250],
    [150, 180, 210],
    [300, 330, 360]
])

q2 = torch.tensor([
    [209, 231, 260],
    [155, 192, 240],
    [302, 338, 390]
])
```

```python
q1 + q2
```

```python
q2 - q1
```

```python
(q2 - q1) * 100 /q1
```

```python
return_rate = 0.1
return_rate * q1
```

```python
profit_per_unit = torch.tensor([
    [30, 27, 44],
    [22, 16, 41],
    [11, 22, 33]
])
```

```python
profit_per_unit * q1
```

```python
product_prices = torch.tensor([[1100, 650, 3200]])
product_prices
```

```python
torch.matmul(product_prices, q1)
```

```python
product_prices
```

```python
product_prices.t()  # Transpose
```

```python
prices_by_region = q1 * product_prices.t()
prices_by_region
```

```python
prices_by_region.sum(dim=0) #column wise sum
```
