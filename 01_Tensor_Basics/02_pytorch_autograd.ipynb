#### Autograd in PyTorch

```python
import torch
```

```python
material_cost = torch.tensor(10, requires_grad=True, dtype=torch.float16)
labour_cost = torch.tensor(5, requires_grad=True, dtype=torch.float16)

total_cost = 3 * material_cost ** 2 + 5 * labour_cost + 100
total_cost
```

```python
total_cost.backward()
material_cost.grad, labour_cost.grad
```

```python
x = torch.tensor(4, requires_grad=True, dtype=torch.float16)
y = x**2 + 5
print("Outside no grad", y.requires_grad)
y.backward()
x.grad
```

```python
x = torch.tensor(4, requires_grad=True, dtype=torch.float16)

with torch.no_grad():
    y = x**2 + 5
    print("Inside no grad", y.requires_grad)
    # y.backward()
```

```python
x = torch.tensor([[1, 2], [3, 4]], requires_grad=True, dtype=torch.float16)
y = 2 * x ** 3 + 7
y
```

```python
# derivative = 6x^2
result = y.sum()
result.backward()
```
